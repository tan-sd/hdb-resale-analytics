#### Effect of: Electoral boundary, ruling party, opposition vs pap 500m proximity

```python
# find unique electoral boundaries
raw_data["Electoral Boundary"].unique()
```

```python
import geopandas as gpd
from shapely.geometry import Point

df = raw_data[raw_data["Year"].between(2006, 2023)].copy()

def get_period(year):
    if 2006 <= year <= 2010:
        return "2006–2010"
    elif 2011 <= year <= 2012:
        return "2011–2012"
    elif 2013 <= year <= 2014:
        return "2013–2014"
    elif 2015 <= year <= 2019:
        return "2015–2019"
    elif 2020 <= year <= 2023:
        return "2020–2023"
    return "Other"

df["Period"] = df["Year"].apply(get_period)

df["geometry"] = df.apply(lambda row: Point(row["Longitude"], row["Latitude"]), axis=1)
gdf = gpd.GeoDataFrame(df, geometry="geometry", crs="EPSG:4326").to_crs(epsg=3414)

wp = gdf[gdf["Ruling Party"] == "WP"].copy()
pap = gdf[gdf["Ruling Party"] == "PAP"].copy()
sda = gdf[gdf["Ruling Party"] == "SDA"].copy()
```

```python
from shapely.ops import unary_union
# buffer each WP/SDA flat
opp = pd.concat([wp, sda])
opp_buffered = opp.copy()
opp_buffered["geometry"] = opp_buffered.geometry.buffer(500)

# merge all opposition buffers into one polygon
merged_buffer = unary_union(opp_buffered.geometry)

# check which PAP flats fall inside polygon
pap["Zone"] = pap.geometry.within(merged_buffer)
pap_near_opp = pap[pap["Zone"]].copy()
pap_near_opp["Zone"] = "PAP"

opp["Zone"] = opp["Ruling Party"]
border_flats = pd.concat([pap_near_opp, opp])
```

```python
plt.figure(figsize=(14, 6))
sns.boxplot(data=border_flats, x="Period", y="Resale Price Adj 2024", hue="Zone")
plt.title("Resale Prices (2024 Adj) Within 500m of WP/SDA–PAP Boundaries")
plt.xlabel("Electoral Period")
plt.ylabel("Resale Price (2024 Adjusted)")
plt.legend(title="Ruling Party")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# find % of WP flats in Sengkang from 2020-2023
wp_2020_2023 = border_flats[
    (border_flats["Ruling Party"] == "WP") &
    (border_flats["Year"].between(2020, 2023))
]

sengkang_count = (wp_2020_2023["Planning Area"] == "SENGKANG").sum()
total_wp_flats = wp_2020_2023.shape[0]
percentage = sengkang_count / total_wp_flats * 100 if total_wp_flats > 0 else 0

print(f"WP flats in Sengkang (2020–2023): {sengkang_count} / {total_wp_flats} ({percentage:.2f}%)")
```

```python
wp_2020_2023 = border_flats[
    (border_flats["Ruling Party"] == "WP") &
    (border_flats["Year"].between(2020, 2023))
]

wp_median_prices = (
    wp_2020_2023.groupby("Planning Area")["Resale Price Adj 2024"]
    .median()
    .sort_values(ascending=False)
    .reset_index()
)

print(wp_median_prices)
```

#### Demographic age group charts

```python
pop_df = pd.read_csv("../population_demographics/age_group_hdb_population_by_year.csv")
pop_df.set_index("Age Group", inplace=True)

age_bin_map = {
    "0-4 Years": "0–24", "5-9 Years": "0–24",
    "10-14 Years": "0–24", "15-19 Years": "0–24",
    "20-24 Years": "0–24", "25-29 Years": "25–59",
    "30-34 Years": "25–59", "35-39 Years": "25–59",
    "40-44 Years": "25–59", "45-49 Years": "25–59",
    "50-54 Years": "25–59", "55-59 Years": "25–59",
    "60-64 Years": "60+", "65-69 Years": "60+",
    "70-74 Years": "60+", "75-79 Years": "60+",
    "80-84 Years": "60+", "85-89 Years": "60+", "90 Years & Over": "60+"
}

pop_df['Age Bin'] = pop_df.index.map(age_bin_map)
binned_pop_df = pop_df.groupby("Age Bin").sum()

file_info = [
    ("../population_demographics/age_and_dwelling_1and2_room_flats.csv", "1- & 2-Room Flats"),
    ("../population_demographics/age_and_dwelling_3and4_room_flats.csv", "3- & 4-Room Flats"),
    ("../population_demographics/age_and_dwelling_5room_and_exec_flats.csv", "5-Room & Executive Flats")
]

pct_by_flat_type = {}
for filepath, label in file_info:
    df = pd.read_csv(filepath)
    df.set_index("Age Group", inplace=True)
    df['Age Bin'] = df.index.map(age_bin_map)

    binned_flat_df = df.groupby("Age Bin").sum()
    # % of each age group living in flat type
    pct_df = (binned_flat_df / binned_pop_df) * 100
    pct_by_flat_type[label] = pct_df
    
flat_1_2_df = pct_by_flat_type["1- & 2-Room Flats"].copy()
flat_3_4_df = pct_by_flat_type["3- & 4-Room Flats"].copy()
flat_5_exec_df = pct_by_flat_type["5-Room & Executive Flats"].copy()
```

```python
def plot_age_bins(df, title):
    plt.figure(figsize=(10, 6))

    for age_bin in df.index:
        plt.plot(df.columns.astype(int), df.loc[age_bin], label=age_bin, marker='o')

    plt.title(f"% of Age Group Living in {title}")
    plt.xlabel("Year")
    plt.ylabel("Percentage of Age Group (%)")
    plt.legend(title="Age Bin")
    plt.tight_layout()
    plt.show()

plot_age_bins(flat_1_2_df, "1- & 2-Room Flats")
plot_age_bins(flat_3_4_df, "3- & 4-Room Flats")
plot_age_bins(flat_5_exec_df, "5-Room & Executive Flats")
```

```python
age_bin_colors = {
    "0–24": 'blue', "25–59": 'orange', "60+": 'green'
}

def plot_all_age_bins(flat_1_2_df, flat_3_4_df, flat_5_exec_df):
    plt.figure(figsize=(12, 8))

    # Plot 1- & 2-Room Flats
    for age_bin in flat_1_2_df.index:
        plt.plot(flat_1_2_df.columns.astype(int), flat_1_2_df.loc[age_bin], label=f"1- & 2-Room Flats - {age_bin}", marker='o', color=age_bin_colors[age_bin])

    # Plot 3- & 4-Room Flats
    for age_bin in flat_3_4_df.index:
        plt.plot(flat_3_4_df.columns.astype(int), flat_3_4_df.loc[age_bin], label=f"3- & 4-Room Flats - {age_bin}", marker='x', color=age_bin_colors[age_bin])

    # Plot 5-Room & Executive Flats
    for age_bin in flat_5_exec_df.index:
        plt.plot(flat_5_exec_df.columns.astype(int), flat_5_exec_df.loc[age_bin], label=f"5-Room & Executive Flats - {age_bin}", marker='s', color=age_bin_colors[age_bin])

    plt.title("Percentage of Age Group Living in Different Flat Types")
    plt.xlabel("Year")
    plt.ylabel("Percentage of Age Group (%)")
    plt.legend(title="Age Bin and Flat Type")
    plt.tight_layout()
    plt.show()

# Call the function with your data
plot_all_age_bins(flat_1_2_df, flat_3_4_df, flat_5_exec_df)
```

```python
df = pd.read_csv("../population_demographics/demographics-segregated-long.csv")

df_age_group = df[df['Category'] == 'age_group']
df_age_group['Value'] = pd.to_numeric(df['Value'], errors='coerce')
df_age_group['Value'] = df_age_group['Value'].fillna(0).astype(int) 
df_age_group['Attribute'] = df['Attribute'].astype(str)
df_age_group = df_age_group[~df_age_group['Attribute'].str.contains('total', na=False)]
df_age_group = df_age_group.groupby(['Year', 'Planning Area', 'Attribute', 'Gender'], as_index=False)['Value'].sum()

def categorize_age_group(row):
    if 'age_0_4' in row['Attribute'] or 'age_5_9' in row['Attribute'] or 'age_10_14' in row['Attribute'] or 'age_15_19' in row['Attribute'] or 'age_20_24' in row['Attribute']:
        return '0-24'
    elif 'age_25_29' in row['Attribute'] or 'age_30_34' in row['Attribute'] or 'age_35_39' in row['Attribute']:
        return '25-39'
    elif 'age_40_44' in row['Attribute'] or 'age_45_49' in row['Attribute'] or 'age_50_54' in row['Attribute'] or 'age_55_59' in row['Attribute']:
        return '40-59'
    else:
        return '60+'

df_age_group['Age Range'] = df_age_group.apply(categorize_age_group, axis=1)
grouped_df = df_age_group.groupby(['Year', 'Planning Area', 'Age Range'], as_index=False)['Value'].sum()
grouped_df = grouped_df.sort_values(by=['Year', 'Planning Area', 'Age Range'], ascending=[True, True, True])

grouped_df
```
